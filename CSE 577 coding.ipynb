```python
#common operations on vectors

import numpy as np
vector_a =np.array([2,4])
vector_b = np.array([6,9])
vector_c = vector_a + vector_b
print("final vector", vector_c)
```

```python
#common operations on matrices

import numpy as np
matrix_a =np.array([[2,4],[6,7]])
matrix_b =np.array([[6,9],[8,9]])
matrix_c = matrix_a + matrix_b
print ("final matrix", matrix_c)
```

```python
#dot product of vector

import numpy as np
vector_a = np.array([2,3])
vector_b = np.array([4,5])
dot_product = np.dot(vector_a, vector_b)
print("final result:", dot_product)
```

```python
# 2x+y=5
# 3x-y = 4

import numpy as np
A= np.array([[2,1],[3,-1]])
B= np.array([5,4])
x,y = np.linalg.solve(A,B)
print("The value of x:",x)
print("The value of y:",y)
```
