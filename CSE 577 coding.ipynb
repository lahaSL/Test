```python
#Practical No. 1
# Vector and matrix operations on traffic data
#Construct mathematical vector and matrix representations from emperical data samples of a city's dailt traffic flow across different intersections, and execute fundamental linear algebraic operations using Python to calculate the total congestion index for each district.

#If a city has intersections and disttricts
#Then we can create
#tarffic vector x: (traffic volume at each intersection)
#District intersection matrix A: (which intersection belong to which district)
#Congestion index vector C: (Total Congestion)

#Solutions: Let x be n x 1 matrix
#Let A be m x n matrix
#Congestion index C = AX
#Let x =[120,80,150] , A = [[1,0,1][0,1,1]]
#C1= 1(120) +0(80)+1(150) = 270
#C2=0(120)+1(80)+1(150) = 230
#C=[270,230]
```

```python
import numpy as np
#traffic at  3 intersections
x=np.array([120,80,150])
#district intersection mapping
A=np.array([[1,0,1],[0,1,1]])

#Congestion index
C = A@x
print("Traffic Vector:")
print(x)

print("District-Intersection Matrix:")
print(A)

print("Congestion index:")
print(C)
```

```python
#Scaling in Linear Transformation

import numpy as np
#scaling matrix
scale_factor = 2
scaling_matrix = np.array([[scale_factor,0],[0,scale_factor]])

#original vector
vector = np.array([1,2])

#applying scaling
scaled_vector = scaling_matrix @ vector
print("scaled vector:", scaled_vector)
```

```python
#Rotation in linear transformation
import numpy as np
#rotation angle (in radians)
theta=np.pi/4  #45 degree
rotation_matrix=np.array([[np.cos(theta), -np.sin(theta)],[np.sin(theta),np.cos(theta)]])
#original vector
vector=np.array([1,0])
#applying rotation
rotated_vector=rotation_matrix @vector
print("rotated vector",rotated_vector)
```

```python
#refelection
import numpy as np
#reflection matrix (y axis)
reflection_matrix=np.array([[1,0],[0,-1]])

#original metrix
vector=np.array([1,2])
#applying reflection
reflected_vector=reflection_matrix@vector
print("reflected vector",reflected_vector)
```

```python
#Euclidean Norm
import numpy as np
#define the vector
vector =np.array([3,4])

#Calculate the Euclidean norm
norm = np.linalg.norm(vector)
print("Euclidean norm:",norm)
```

```python
#Manhattan Norm
import numpy as np
#define the vector
vector =np.array([3,4])

#Calculate the Euclidean norm
norm = np.linalg.norm(vector,ord=1)
print("Manhattan norm:",norm)
```

```python
#infinity notm
norm = np.linalg.norm(vector,ord=np.inf)
print("infinity norm:",norm)
```

```python
#In linear algebra the concept of projection is used to determine the how much of one vector lies in a direction of another vector.
```

```python
import numpy as np

# Drone position
p = np.array([1, 0])

# Rotation matrix for 90 degrees
theta = np.pi / 2
R = np.array([[np.cos(theta), -np.sin(theta)], 
              [np.sin(theta),  np.cos(theta)]])

# Rotate position
p_rotated = R @ p

print("R:\n", np.round(R, 4))
print("Rotated p:", np.round(p_rotated, 4))
```
