```python
# 1. Loss and Gradient Descent
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
# 1. Inputs and Parameters (Must be floating point for GradientTape)
x = tf.constant([1.0, 2.0, 3.0], dtype=tf.float32)
y = tf.constant([2.0, 4.0, 5.0], dtype=tf.float32)
m = tf.Variable(0.0, dtype=tf.float32)
c = tf.Variable(0.0, dtype=tf.float32)
lr = 0.1  # learning rate (lowercase L)
epoch = 20
# 2. Loss and Gradient Descent Loop
history = []
for i in range(epoch + 1):
    with tf.GradientTape() as tape:
        pred = m * x + c
        loss = tf.reduce_mean((y - pred) ** 2)
    # Calculate gradients outside the GradientTape block
    dm, dc = tape.gradient(loss, [m, c])
    history.append([i, m.numpy(), c.numpy(), loss.numpy(), dm.numpy(), dc.numpy()])
    if i < epoch:
        m.assign(m - lr * dm)
        c.assign(c - lr * dc)
# 3. Display Results Table
df = pd.DataFrame(history, columns=["Iteration", "m", "c", "loss", "dm", "dc"])
df.round(4)


```
