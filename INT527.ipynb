```python
#different seed values (any non-negative integer) affect the random output

import tensorflow as tf
#using seed = 1
gen1=tf.random.Generator.from_seed(1)
print("seed 1:",gen1.normal(shape=[3]))
#Using seed = 42
gen42 = tf.random.Generator.from_seed(42)
print("seed 42:",gen42.normal(shape=[3]))

#Using seed = 123
gen42 = tf.random.Generator.from_seed(123)
print("seed 123:",gen42.normal(shape=[3]))
```
