```python
#Define Matrix Tensor

matrix_tensor =tf.constant([[1,2],[3,4]])
print(matrix_tensor)

#Define a ones tensor creates a tensor of all ones that has the same shape as the input
ones_tensor=tf.ones_like(matrix_tensor)
print(ones_tensor)

#Element wise addition
tensor_add =tf.add(matrix_tensor,ones_tensor)
print(tensor_add)

#Matrix multiplication (dot product)

matrix_mult =tf.matmul(matrix_tensor,matrix_tensor)

print(matrix_mult)

```

```python
#Transpose

x=tf.constant([[1,2],[3,4],[5,6]])
y=tf.constant([[7,8],[9,10],[11,12]])
print(tf.transpose(x))
```

```python
z = tf.transpose(x)
print(tf.tensordot(z, y, axes=0))
```

```python
# Mixed precision => float16 runs faster than float32; Update datatype as:
d = tf.constant([10, 7], dtype=tf.float16)
print(d)
print(d.dtype)
```

```python
# 1. Define an int32 tensor (default for integer literals)
c = tf.constant([10, 7])
print("Original:", c, "dtype:", c.dtype)

# 2. Change from int32 to float32
E = tf.cast(c, dtype=tf.float32)
print("Casted:", E, "dtype:", E.dtype)
```

```python
# Get absolute values
d = tf.constant([-7, -10, 5, -3])
abs_d = tf.abs(d)

print("Original tensor:", d)
print("Absolute values:", abs_d)
```
